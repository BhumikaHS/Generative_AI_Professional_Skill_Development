```python
import time
import uuid
from datetime import datetime
```

```python
class Blackboard:

    def __init__(self):

        self.memory = {}

        self.transactions = []


    def write(
        self,
        agent,
        key,
        value
    ):

        transaction_id = str(
            uuid.uuid4()
        )
        record = {
            "transaction_id": transaction_id,
            "agent": agent,
            "key": key,
            "value": value,
            "timestamp": datetime.now()
        }

        self.memory[key] = value

        self.transactions.append(
            record
        )

        print(
            f"[{agent}] wrote: {key}"
        )

        return transaction_id


    def read(self, key):

        return self.memory.get(
            key,
            None
        )

    def show_transactions(self):

        print("\nTRANSACTION LOG")

        for transaction in self.transactions:

            print(
                transaction
            )
```

```python
# 2. CREATE BLACKBOARD

blackboard = Blackboard()

# 3. RESEARCH AGENT

class ResearchAgent:

    def run(self, problem):

        result = (
            "Research Agent: "
            "The system should use AI, "
            "robotics and predictive analytics."
        )

        blackboard.write(
            "ResearchAgent",
            "research_result",
            result
        )
```

```python
# 4. SYSTEM DESIGN AGENT

class SystemAgent:

    def run(self):

        research = blackboard.read(
            "research_result"
        )

        architecture = (
            "System Agent: "
            "Architecture consists of "
            "sensor layer, AI layer, "
            "robot control layer and "
            "monitoring layer."
        )

        blackboard.write(
            "SystemAgent",
            "architecture",
            architecture
        )
```

```python
# 5. AUDITOR AGENT

class AuditorAgent:

    def run(self):

        architecture = blackboard.read(
            "architecture"
        )

        audit = (
            "Auditor Agent: "
            "Architecture reviewed. "
            "Safety, reliability and "
            "monitoring should be included."
        )

        blackboard.write(
            "AuditorAgent",
            "audit_result",
            audit
        )
```

```python
# 6. FINAL SYNTHESIS AGENT

class SynthesisAgent:

    def run(self):

        research = blackboard.read(
            "research_result"
        )

        architecture = blackboard.read(
            "architecture"
        )

        audit = blackboard.read(
            "audit_result"
        )

        final_report = f"""
FINAL MULTI-AGENT REPORT
Research:
{research}

Architecture:
{architecture}

Audit:
{audit}
"""

        blackboard.write(
            "SynthesisAgent",
            "final_report",
            final_report
        )
```

```python
# 7. RUN AGENTS

problem = (
    "Design an AI-powered smart factory."
)

research_agent = ResearchAgent()
system_agent = SystemAgent()
auditor_agent = AuditorAgent()
synthesis_agent = SynthesisAgent()


research_agent.run(
    problem
)

system_agent.run()

auditor_agent.run()

synthesis_agent.run()
```

```python
# 8. DISPLAY FINAL RESULT
print("\n")
print(
    blackboard.read(
        "final_report"
    )
)

# 9. TRANSACTION LOG

blackboard.show_transactions()
```

```python
import matplotlib.pyplot as plt

agents = [
    "Research",
    "System",
    "Auditor",
    "Synthesis"
]

transactions = [
    1,
    2,
    3,
    4
]

plt.figure(figsize=(9, 5))

plt.plot(
    agents,
    transactions,
    marker="o"
)

plt.xlabel("Agent")
plt.ylabel("Transaction Number")

plt.title(
    "Multi-Agent Blackboard Transactions"
)

plt.grid()

plt.show()
```
