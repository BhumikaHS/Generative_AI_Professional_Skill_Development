**OBJECTIVES**

Build a simplified Enterprise LLM Gateway that acts as a single
interface between users and different Large Language Models (LLMs).

The gateway demonstrates:

- Request validation
- Authentication
- Rate limiting
- Model routing
- Token usage tracking
- Logging
- Error handling
- Response generation

In an enterprise environment, an LLM gateway can provide a centralized
layer for controlling and monitoring access to multiple AI models.

```python
import time
from collections import defaultdict

print("Libraries imported successfully!")
```

**Define Available LLM Models**

An enterprise gateway may provide access to multiple models.

For this simplified implementation, we create three simulated models:

- small-model
- medium-model
- large-model

These are simulated locally, so no API key is required.

```python
models = {
    "small-model": {
        "cost": 0.001,
        "description": "Fast model for simple tasks"
    },

    "medium-model": {
        "cost": 0.003,
        "description": "Balanced model"
    },

    "large-model": {
        "cost": 0.01,
        "description": "Powerful model for complex tasks"
    }
}

print("Available models:")

for name, details in models.items():
    print(
        name,
        "->",
        details["description"]
    )
```

**Authentication**

The gateway checks whether the user has a valid API key.

This is a simplified demonstration of enterprise authentication.

```python
api_keys = {
    "user_001": "key-demo-001",
    "user_002": "key-demo-002",
    "admin": "key-admin-001"
}

print("Registered users:", len(api_keys))
```

```python
def authenticate(user_id, api_key):

    if user_id not in api_keys:
        return False

    if api_keys[user_id] != api_key:
        return False

    return True
```

**Test Authentication**

The gateway should allow valid users and reject invalid credentials.

```python
print(
    "Valid login:",
    authenticate(
        "user_001",
        "key-demo-001"
    )
)

print(
    "Invalid login:",
    authenticate(
        "user_001",
        "wrong-key"
    )
)
```

**Rate Limiting**

Enterprise systems need to prevent a single user from sending unlimited
requests.

We create a simple request limit.

Each user can make a maximum of 5 requests.

```python
request_limit = 5

request_count = defaultdict(int)

def check_rate_limit(user_id):

    if request_count[user_id] >= request_limit:
        return False

    request_count[user_id] += 1

    return True
```

```python
test_user = "user_001"

for i in range(7):

    allowed = check_rate_limit(
        test_user
    )

    print(
        f"Request {i + 1}:",
        "Allowed" if allowed else "Blocked"
    )
```

**Reset Rate Limit**

For the remaining experiments, we reset the request counter.

```python
request_count.clear()

print("Rate limits reset.")
```

**Model Routing**

The gateway can select a model depending on the complexity of a request.

For this simplified system:

- Simple → small-model
- Medium → medium-model
- Complex → large-model

```python
def select_model(complexity):

    if complexity == "simple":
        return "small-model"

    elif complexity == "medium":
        return "medium-model"

    elif complexity == "complex":
        return "large-model"

    else:
        return None
```

```python
complexities = [
    "simple",
    "medium",
    "complex"
]

for complexity in complexities:

    model = select_model(
        complexity
    )

    print(
        complexity,
        "->",
        model
    )
```

**Token Usage Tracking**

Enterprise applications often track token usage for monitoring and
billing.

For simplicity, we estimate tokens using the number of words.

```python
def estimate_tokens(text):

    return len(
        text.split()
    )
```

```python
text = "Explain generative AI and large language models."

tokens = estimate_tokens(
    text
)

print(
    "Estimated tokens:",
    tokens
)
```

**Calculate Request Cost**

Each model has a different cost per token.

The gateway calculates the estimated cost of each request.

```python
def calculate_cost(
    model_name,
    tokens
):

    cost_per_token = models[
        model_name
    ]["cost"]

    return tokens * cost_per_token
```

**Simulated LLM Response**

We do not need an external LLM API for this project.

Instead, we create a simulated response so that the gateway logic can
be demonstrated without API keys.

```python
def generate_response(
    model_name,
    prompt
):

    return (
        f"[{model_name}] "
        f"Response generated for: "
        f"{prompt}"
    )
```

**Request Logging**

The gateway records important information about each request.

The log contains:

- User
- Model
- Tokens
- Cost
- Status

```python
request_logs = []

def log_request(
    user_id,
    model,
    tokens,
    cost,
    status
):

    request_logs.append({
        "user": user_id,
        "model": model,
        "tokens": tokens,
        "cost": cost,
        "status": status
    })
```

**Enterprise LLM Gateway**

Now we combine all components into one gateway.

The gateway performs:

1. Authentication
2. Rate limiting
3. Model selection
4. Token estimation
5. Cost calculation
6. Response generation
7. Logging

```python
def llm_gateway(
    user_id,
    api_key,
    prompt,
    complexity
):

    # Authentication
    if not authenticate(
        user_id,
        api_key
    ):

        return {
            "status": "ERROR",
            "message": "Authentication failed"
        }

    # Rate limiting
    if not check_rate_limit(
        user_id
    ):

        return {
            "status": "ERROR",
            "message": "Rate limit exceeded"
        }

    # Model routing
    model = select_model(
        complexity
    )

    if model is None:

        return {
            "status": "ERROR",
            "message": "Invalid complexity"
        }

    # Token estimation
    tokens = estimate_tokens(
        prompt
    )

    # Cost
    cost = calculate_cost(
        model,
        tokens
    )

    # Response
    response = generate_response(
        model,
        prompt
    )

    # Logging
    log_request(
        user_id,
        model,
        tokens,
        cost,
        "SUCCESS"
    )

    return {
        "status": "SUCCESS",
        "model": model,
        "tokens": tokens,
        "cost": cost,
        "response": response
    }
```

**Send an LLM Request**

We now send a request through the complete enterprise gateway.

```python
result = llm_gateway(
    user_id="user_001",
    api_key="key-demo-001",
    prompt="Explain retrieval augmented generation.",
    complexity="medium"
)

print("Status:", result["status"])
print("Model:", result["model"])
print("Tokens:", result["tokens"])
print("Cost:", result["cost"])
print("Response:", result["response"])
```

**Test Authentication Failure**

The gateway should reject requests with invalid API credentials.

```python
result = llm_gateway(
    user_id="user_001",
    api_key="wrong-key",
    prompt="Explain machine learning.",
    complexity="simple"
)

print(result)
```

**Test Model Routing**

We send requests with different complexity levels and observe which
model is selected.

```python
request_count.clear()

for complexity in [
    "simple",
    "medium",
    "complex"
]:

    result = llm_gateway(
        user_id="user_002",
        api_key="key-demo-002",
        prompt="Explain artificial intelligence.",
        complexity=complexity
    )

    print(
        complexity,
        "->",
        result["model"]
    )
```

**View Request Logs**

The gateway maintains a record of successful requests.

```python
print("========== REQUEST LOGS ==========")

for log in request_logs:

    print(log)

print("==================================")
```

**Usage Statistics**

We calculate total tokens and estimated cost across all successful
requests.

```python
total_tokens = sum(
    log["tokens"]
    for log in request_logs
)

total_cost = sum(
    log["cost"]
    for log in request_logs
)

print(
    "Total tokens:",
    total_tokens
)

print(
    "Total estimated cost:",
    round(
        total_cost,
        4
    )
)
```
